# WSL Miniconda 환경 설정

> 아래 명령어는 WSL Ubuntu 터미널에서 실행합니다.  
> Miniconda 설치와 Conda 명령에는 `sudo`를 사용하지 않습니다.

## 1. 설치 파일 확인

```bash
ls -lh Miniconda3-latest-Linux-x86_64.sh
uname -m
````

`uname -m` 결과가 `x86_64`이면 현재 설치 파일을 사용할 수 있습니다.

## 2. Miniconda 설치

```bash
bash Miniconda3-latest-Linux-x86_64.sh
```

### 설치 과정

* 라이선스 확인: `Enter`
* 라이선스 동의: `yes`
* 기본 설치 경로 사용: `Enter`
* Conda 초기화: `yes`

## 3. 설정 적용

```bash
source ~/.bashrc
```

## 4. 설치 확인

```bash
conda --version
which conda
conda info
```

## 5. Conda 업데이트

```bash
conda update -n base conda -y
```

## 6. 가상환경 생성

```bash
conda create -n nvidia python=3.11 -y
conda activate nvidia
```

### 환경 확인

```bash
python --version
conda env list
```

## 7. 패키지 및 Jupyter 설치

```bash
conda install pandas matplotlib seaborn numpy scikit-learn jupyterlab ipykernel -y
pip install koreanize-matplotlib
```

### 설치 확인 및 실행

```bash
conda list
jupyter --version
jupyter lab
```

