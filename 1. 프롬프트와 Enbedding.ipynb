## 코사인 유사도
- 측정 방식: 두 벡터 사이의 코사인 각도를 계산.
- 결과 범위: 값은 -1에서 1 사이의 값을 가진다
  - 1에 가까울수록 두 벡터는 같은 방향(매우 유사)
  - 0에 가까울수록 직교(연관성 없음)
  - 1에 가까울수록 반대 방향(정반대)

```python
# 유사도 검색 -> 코사이 유사도

import numpy as np
from sklearn.metrics.pairwise  import cosine_similarity

v1 = np.array([[1, 2, 3]])
v2 = np.array([[10, 20, 30], [-5, -6, -7]])

cos_sim = cosine_similarity(v1, v2)
print(cos_sim)
```

### 무료: SentenceTransformer
- 텍스트(문장, 단락, 문단 등)를 고정된 크기의 밀집 벡터(Dense Vector)로 변환해 주는 파이썬 라이브러리
- 단순히 단어의 빈도를 세는 방식(TF-IDF 등)과 달리, 문장의 의미적 맥락을 파악하여 벡터로 표현하는 것이 특징
- 주요 특징
  - 의미적 유사도(Semantic Similarity): "점심 메뉴로 무엇이 좋을까?"와 "오늘 점심은 뭘 먹지?"라는 두 문장은 단어 구성이 완전히 다르지만, 의미는 같습니다. SentenceTransformer는 이 두 문장을 비슷한 위치의 벡터로 변환하여, 코사인 유사도 계산 시 높은 유사도를 보여줍니다.
  - 임베딩 벡터 생성: 텍스트를 고차원 숫자 배열로 변환함으로써, 벡터 데이터베이스(Vector DB)에 저장하여 빠른 검색이 가능하게 합니다. 이는 최근의 RAG(Retrieval-Augmented Generation) 시스템 구축의 핵심 요소입니다.
  - 간편한 사용: Hugging Face에 공개된 수많은 모델을 단 몇 줄의 코드로 불러와서 즉시 사용할 수 있습니다.
- 사용 이유
  - 고차원 표현: 기존의 단순 벡터화 방식보다 훨씬 풍부한 의미 정보(고차원)를 담고 있습니다.
  - 다양한 언어 지원: 한국어 데이터로 특화 학습된 모델(ko-sroberta 등)을 사용하면 한국어의 미묘한 의미 차이까지 잘 잡아낼 수 있습니다.

```python
# 무료 버전을 이용한 벡터화
from sentence_transformers import SentenceTransformer

# 1. 모델 불러오기
model = SentenceTransformer('jhgan/ko-sroberta-multitask')

# 2. 임베딩 생성할 문장들 => 문장 사전
sentences = [
    "고양이가 창가에서 햇볕을 쬐고 있다",
    "새끼 고양이가 실타래를 가지고 논다",
    "강아지가 공원에서 뛰어다닌다",
    "강아지가 주인을 보고 꼬리를 흔든다",
    "호랑이가 숲속을 걸어 다닌다",
    "사자가 초원에서 사냥을 준비한다",
    "코끼리가 긴 코로 나뭇가지를 잡는다",
    "기린이 나무 위 잎을 먹는다",
    "원숭이가 바나나를 먹는다",
    "펭귄이 얼음 위에서 걸어간다",
    "사과가 빨갛게 익었다",
    "바나나가 노랗게 잘 익어 달콤하다",
    "포도가 달콤하고 신선하다",
    "빵이 오븐에서 갓 구워졌다",
    "라면이 뜨겁게 끓고 있다",
    "김치찌개에서 얼큰한 냄새가 난다",
    "초밥이 접시에 예쁘게 놓여 있다",
    "피자가 치즈로 가득 덮여 있다",
    "햄버거에 고기 패티가 두껍다",
    "커피 향이 방 안에 가득 퍼졌다",
    "서울의 남산타워가 빛나고 있다",
    "한강 다리 위에서 야경을 본다",
    "제주도의 바닷가가 파랗게 빛난다",
    "부산 해운대 해수욕장에 사람이 많다",
    "경복궁의 전통 건물이 웅장하다",
    "뉴욕의 자유의 여신상이 우뚝 서 있다",
    "파리의 에펠탑이 밤에 반짝인다",
    "런던의 빅벤 시계탑이 울린다",
    "도쿄의 시부야 거리가 붐빈다",
    "로마의 콜로세움이 고대의 흔적을 보여준다"
]
```

```python
# 3. embedding 진행 및 결과 확인
embeddings = model.encode(sentences)

embeddings
```

```python
print(f'인벤팅 벡터수: {len(embeddings[0])}, 차원 : {embeddings.shape}')
print(len(sentences),'\n')
print(embeddings[0])
```

### embedding vector 결과(위치) 확인하기
- TensorFlow Embedding Projector(https://projector.tensorflow.org/): 구글에서 제공하는 웹 사이트를 통해 직접 시각화
  - 1단계: Colab에서 생성한 embeddings 배열을 tsv 파일로 저장
  - 2단계: Embedding Projector 접속 후, 좌측 상단의 [Load] 버튼을 클릭
  - 3단계: Step 1에서 생성한 vectors.tsv를 Step 1(Load a TSV file of vectors)에, metadata.tsv를 Step 2(Load a TSV file of metadata)에 각각 업로드
  - 4단계: 우측의 'T-SNE'나 'PCA'를 선택하여 데이터가 어떻게 뭉쳐 있는지 확인


```python
# 4. vectors.tsv 저장
with open('vectors.tsv', 'w', encoding='utf-8') as f:
    for emb in embeddings:
        f.write('\t'.join([str(x) for x in emb])+ '\n')

# 5. metadata.tsv 저장
with open('metadata.tsv', 'w', encoding='utf-8') as f:
    for s in sentences:
        f.write(s + '\n')
```

```python
# 6. 결과 확인

# 사용자 문장 입력
query = input("문장을 입력하세요.")

# 문장 임베딩
query_embedding = model.encode([query])

# 코사인 유사도 계산
similarity = cosine_similarity(query_embedding, embeddings)[0]

# 유사도가 높은 상위 2개 문장 찾기
top2_idx = similarity.argsort()[-2:][::-1]

print('가장 유사한 문자 2개')
print("="*40)
for idx in top2_idx:
    print(f'문자 : {sentences[idx]}')
    print(f'유사도: {similarity[idx]:.4f}')
    print()
```

```python
# 2. 임베딩 생성할 문장들 => 문장 사전
sentences = [
    "고양이가 창가에서 햇볕을 쬐고 있다",
    "새끼 고양이가 실타래를 가지고 논다",
    "강아지가 공원에서 뛰어다닌다",
    "강아지가 주인을 보고 꼬리를 흔든다",
    "호랑이가 숲속을 걸어 다닌다",
    "사자가 초원에서 사냥을 준비한다",
    "코끼리가 긴 코로 나뭇가지를 잡는다",
    "기린이 나무 위 잎을 먹는다",
    "원숭이가 바나나를 먹는다",
    "펭귄이 얼음 위에서 걸어간다",
    "사과가 빨갛게 익었다",
    "바나나가 노랗게 잘 익어 달콤하다",
    "포도가 달콤하고 신선하다",
    "빵이 오븐에서 갓 구워졌다",
    "라면이 뜨겁게 끓고 있다",
    "김치찌개에서 얼큰한 냄새가 난다",
    "초밥이 접시에 예쁘게 놓여 있다",
    "피자가 치즈로 가득 덮여 있다",
    "햄버거에 고기 패티가 두껍다",
    "커피 향이 방 안에 가득 퍼졌다",
    "서울의 남산타워가 빛나고 있다",
    "한강 다리 위에서 야경을 본다",
    "제주도의 바닷가가 파랗게 빛난다",
    "부산 해운대 해수욕장에 사람이 많다",
    "경복궁의 전통 건물이 웅장하다",
    "뉴욕의 자유의 여신상이 우뚝 서 있다",
    "파리의 에펠탑이 밤에 반짝인다",
    "런던의 빅벤 시계탑이 울린다",
    "도쿄의 시부야 거리가 붐빈다",
    "로마의 콜로세움이 고대의 흔적을 보여준다"
]
```

```python
# 3. OpenAI Embedding 함수 생성
def get_embedding(text):
    res = client.embeddings.create(
        model='text-embedding-3-large',
        input=text
    )
    return res.data[0].embedding
```

```python
# 4. 문장 임베딩 생성
sentence_embeddings = [get_embedding(s) for s in sentences]
```

```python
# 5. 사용 질문 입력 및 임베딩 생성
query = input('문장을 입력하세요')

query_embedding = get_embedding(query)
```

```python
# 6. 유사도 계산
similarity = cosine_similarity(
    [query_embedding],
    sentence_embeddings
)[0]

# 유사도가 높은 상위 2개 문장 찾기
top2_idx = similarity.argsort()[-2:][::-1]

print('가장 유사한 문자 2개')
print(f'사용자 입력: {query}')
print("="*40)
for idx in top2_idx:
    print(f'문자 : {sentences[idx]}')
    print(f'유사도: {similarity[idx]:.4f}')
    print()
```

### Vertor Store에 저장하기

- 일반적으로 단순힌 vector만 저장하는게 아니라 metadata도 같이 저장
- metadata는 어떤 파일에서 나온건지..(a.txt)
- 페이지 번호, chunk번호 이런것들이 metadata로 저장</b>
</br>
- metadata는 vector store마다 저장 방식이 조금씩 다름
- Chroma(로컬기반), Pinecone(클라우드 기반) => embedding vector와 metadata를 같이 저장
- FAISS(Chroma 기반): metadata를 별도로 저장(페이스북에서 개발)

- Vector Store
  1. Pinecone(파이콘) : 유료(5개까지 무료), 클라우드 기반으로 접근성이 좋음, scale-out(확장성)이 좋음
  2. Chroma(크로마) : 무료, 사용하기 쉬움(사용자 친화적)
  3. Weaviate(위비에이트) : 하이브리드 검색 지원(다양한 유사도 검색 지원)
  4. Milvus(밀버스) : 분산형 Vector DB
  5. FAISS : 고성능, 속도에 장점이 있음
  6. pgvector : PostgresSQL을 사용하는 환경에서 특별한 처리 없이 vector 검색이 가능

```python
import chromadb
from openai import OpenAI
from chromadb import PersistentClient

# LLM 객체
client = OpenAI()

# chromadb 클라이언트 생성
chroma_client = chromadb.PersistentClient(
    path='./chroma_db'
)
```

```python
# 임베딩 모델 생성
from chromadb.utils.embedding_functions import SentenceTransformerEmbeddingFunction

ef = SentenceTransformerEmbeddingFunction(
    model_name='snunlp/KR-SBERT-V40K-klueNLI-augSTS'
)
```

```python
# collection 생성
collection = chroma_client.get_or_create_collection(
    name="loc_docs",   # 벡터DB 이름
    embedding_function=ef,  # 임베딩 모델
    metadata={'hnsw:space':'cosine'}
)
```

```python
docs = [
    "파이썬으로 리스트 정렬하기",
    "자바에서 배열 정렬하기",
    "리눅스에서 파일 복사하기"
]
```

```python
# metadata 생성
# 문자에 대한 정보를 기술
metas = [
    {"doc_id": "py_list.pdf", "page": 10, "chunk": 1, "source":"local"},
    {"doc_id": "java_array.pdf", "page": 20, "chunk": 0, "source":"git"},
    {"doc_id": "linux_file.pdf", "page": 5, "chunk": 1, "source":"web"},
]
```

```python
# 문장 ID 생성
ids = [
    "py_list.pdf::p10::c1",
    "java_array.pdf::p20::c0",
    "linux_file.pdf::p5::c1"
]
```

```python
# Chromdb 저장
collection.upsert(ids=ids,
                  documents=docs,
                  metadatas=metas)
```

```python
# 저장 데이터 확인(일부 데이터 확인, 기본 10개)
res = collection.peek()

data1 = {
    'id':res['ids'][0],
    'docs': res['documents'][0],
    'metadata': res['metadatas'][0]
}

print(data1)
```

```python
# 유사문장 검색
query ="파이썬 리스트를 정렬하는 방법에 대해 설명해주세요"

res = collection.query(
    query_texts=query,     # 검색 문장
    n_results=2,           # 상위 추출 건수
    where={'source':'local'} # 검색 조건( 'source': 'local'인 데이터들 중에서)
)

print(res)
```

```python
from langchain_chroma import Chroma
from langchain_openai import OpenAIEmbeddings
from langchain_core.documents import Document
from dotenv import load_dotenv
from pathlib import Path
import os
import shutil
```

```python
ai_key = os.getenv("OPENAI_API_KEY")

def create_docs(sentences_list: list[str]) -> list[Document]:
    '''
    문자열 문장을 랭체인 Document 객체로 변환
    각 문서의 메타데이터 자동 추가
    - id: 문서번호
    - source: 문서 출처
    '''

    documents = []

    for index, sentence in enumerate(sentences_list, start=1):
        # 공백만 있는 데이터는 생략
        if not sentence.strip():
            continue

        document = Document(
            page_content=sentence.strip(),
            metadata={
                "id": index,
                "source": "loc_example"
            }
        )
        documents.append(document)
    return documents
```

```python
# 원문장을 랭채인 Document 객체로 변환
documents = create_docs(sentences)

for doc in documents:
    print(f'내용: {doc.page_content}')
    print(f'메타데이터: {doc.metadata}')
    print('='*20)
```
