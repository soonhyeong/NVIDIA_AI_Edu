```python
import requests

item_id = input('id 입력')

url = f'http://127.0.0.1:8000/item/{item_id}'
res = requests.get(url)

if res.status_code == 200:
    data = res.json()
    print(data)
    print(f"이름: {data['item']['item_name']}")
    print(f"가격: {data['item']['price']}")
else:
    print(f"Error: {res.status_code} - {res.text}")
    data =""
```
