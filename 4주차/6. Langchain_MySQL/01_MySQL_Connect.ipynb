```python
import pymysql

connection = pymysql.connect(
    host = '127.0.0.1',
    user = 'root',
    password = '1234',
    database = 'mysqldb'
)
```

```python
try:
    with connection.cursor() as cursor:
        sql = """
        insert into menu (name, price, description)
        values (%s, %s, %s)
        """
        cursor.execute(sql, ("불고기 피자", 19000, "한국식 불고기 피자 입니다."))

    connection.commit()
    print("메뉴가 추가되었습니다.")

finally:
    connection.close
```

```python
import pandas as pd
with connection.cursor() as cursor:
    sql = 'select * from menu'
    cursor.execute(sql)

    menus = cursor.fetchall()
    print('=== 메뉴 출력 ===')

    for menu in menus:
        print(
           print(menu)
        )
    df = pd.DataFrame(menus)

df
```
