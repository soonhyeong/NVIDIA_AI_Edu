```python
from datetime import datetime
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.runnables import RunnableConfig, chain

# 오늘 날짜 설정
today = datetime.today().strftime("%Y-%m-%d")

# 프롬프트 템플릿 
prompt = ChatPromptTemplate([
    ("system", f"You are a helpful AI assistant. Today's date is {today}."),
    ("human", "{user_input}"),
    ("placeholder", "{messages}"),
])

# ChatOpenAI 모델 초기화 
llm = ChatOpenAI(model="gpt-4o-mini")

# LLM에 도구를 바인딩
llm_with_tools = llm.bind_tools(tools=[web_search])

# 프롬트트와 바인딩 LLM을 LCEL체인으로 연결(LLM 체인 생성)
llm_chain = prompt | llm_with_tools

# 도구 실행 체인 정의
@chain   # 데코레이터 생성
def web_search_chain(user_input: str, config: RunnableConfig):
    
    # 프로프트 내용이 전달
    input_ = {"user_input": user_input}  

    # 처음 실행시 도구호출(툴 콜스 들이 들어옴)
    ai_msg = llm_chain.invoke(input_, config=config)  
    print("ai_msg: \n", ai_msg)
    print("-"*100)

    # 툴 콜드 메세지를 이용해 실제 검색 수행
    tool_msgs = web_search.batch(ai_msg.tool_calls, config=config)  
    print("tool_msgs: \n", tool_msgs)
    print("-"*100)

    return llm_chain.invoke({**input_, "messages": [ai_msg, *tool_msgs]}, config=config)
            #  최종적으로는 사용자 입력, 도구 호출 내용, 도구 호출의 결과 등이 메시지 필드의 플레이스(placeholder) 홀더로 전달

# 체인 실행
response = web_search_chain.invoke("헌법에서 소득세 관련은 어디에 있나요?")

# 응답 출력 
pprint(response.content)
```

```python
from langchain_community.tools import TavilySearchResults
from langchain_core.tools import tool
from typing import List

# Tool 정의 
@tool
def search_web(query: str) -> str:
    """Searches the internet for information that does not exist in the database or for the latest information."""

    tavily_search = TavilySearchResults(max_results=2)
    docs = tavily_search.invoke(query)

    formatted_docs = "\n---\n".join([
        f'<Document href="{doc["url"]}"/>\n{doc["content"]}\n</Document>'
        for doc in docs
        ])

    if len(formatted_docs) > 0:
        return formatted_docs
    
    return "관련 정보를 찾을 수 없습니다."
```

```python
from langchain_community.document_loaders import WikipediaLoader
from langchain_core.documents import Document
from langchain_core.runnables import RunnableLambda
from pydantic import BaseModel, Field
from typing import List

# WikipediaLoader를 사용하여 위키피디아 문서를 검색하는 함수 
def search_wiki(input_data: dict) -> List[Document]:
    """Search Wikipedia documents based on user input (query) and return k documents"""
    query = input_data["query"]
    k = input_data.get("k", 2)  
    wiki_loader = WikipediaLoader(query=query, load_max_docs=k, lang="ko")
    wiki_docs = wiki_loader.load()
    return wiki_docs

# 도구 호출에 사용할 입력 스키마 정의 
class WikiSearchSchema(BaseModel):
    """Input schema for Wikipedia search."""
    query: str = Field(..., description="The query to search for in Wikipedia")
    k: int = Field(2, description="The number of documents to return (default is 2)")

# RunnableLambda 함수를 사용하여 위키피디아 문서 로더를 Runnable로 변환 
runnable = RunnableLambda(search_wiki)
wiki_search = runnable.as_tool(
    name="wiki_search",
    description=dedent("""
        Use this tool when you need to search for information on Wikipedia.
        It searches for Wikipedia articles related to the user's query and returns
        a specified number of documents. This tool is useful when general knowledge
        or background information is required.
    """),
    args_schema=WikiSearchSchema
)
```

```python
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
from langchain_core.runnables import RunnableLambda
from langchain_community.document_loaders import WikipediaLoader

# WikipediaLoader를 사용하여 위키피디아 문서를 검색하고 텍스트로 반환하는 함수 
def wiki_search_and_summarize(input_data: dict):
    wiki_loader = WikipediaLoader(query=input_data["query"], load_max_docs=2, lang="ko")
    wiki_docs = wiki_loader.load()

    formatted_docs =[
        f'<Document source="{doc.metadata["source"]}"/>\n{doc.page_content}\n</Document>'
        for doc in wiki_docs
        ]
    
    return formatted_docs

# 요약 프롬프트 템플릿
summary_prompt = ChatPromptTemplate.from_template(
    "Summarize the following text in a concise manner:\n\n{context}\n\nSummary:"
)

# LLM 및 요약 체인 설정
llm = ChatOpenAI(model="gpt-4o-mini", temperature=0)
summary_chain = (
    {"context": RunnableLambda(wiki_search_and_summarize)}
    | summary_prompt | llm | StrOutputParser() 
)

# 요약 테스트 
summarized_text = summary_chain.invoke({"query":"헌법의 유래"})
pprint(summarized_text)
```

```python
from langchain_community.document_loaders import Docx2txtLoader

LAW_FILE_PATH = "./data/소득세법_20260701.docx"

# DOCX 문서 로드
loader = Docx2txtLoader(LAW_FILE_PATH)
documents = loader.load()

print(f"로드된 원본 Document 수: {len(documents)}")
print(f"문서 전체 글자 수: {len(documents[0].page_content):,}")
print(documents[0].page_content[:500])
```

```python
import re

from langchain_core.documents import Document
from langchain_text_splitters import RecursiveCharacterTextSplitter


def split_law_articles(document: Document) -> list[Document]:
    """
    소득세법 문서를 제1조, 제1조의2 등의 조문 단위로 분할합니다.
    """

    text = document.page_content

    # 제1조, 제1조의2, 제50조 등의 조문 시작 지점 검색
    article_pattern = re.compile(
        r"(?m)^\s*(제\d+조(?:의\d+)?(?:\([^)]+\))?)"
    )

    matches = list(article_pattern.finditer(text))

    if not matches:
        print("조문 형식을 찾지 못했습니다.")
        return [document]

    article_documents = []

    for index, match in enumerate(matches):
        start = match.start()

        if index + 1 < len(matches):
            end = matches[index + 1].start()
        else:
            end = len(text)

        article_content = text[start:end].strip()
        article_heading = match.group(1)

        # 제1조 또는 제1조의2 추출
        article_number_match = re.match(
            r"제\d+조(?:의\d+)?",
            article_heading,
        )

        article_number = (
            article_number_match.group()
            if article_number_match
            else article_heading
        )

        # 괄호 안 조문 제목 추출
        title_match = re.search(
            r"\(([^)]+)\)",
            article_heading,
        )

        article_title = (
            title_match.group(1)
            if title_match
            else ""
        )

        article_document = Document(
            page_content=article_content,
            metadata={
                "source": document.metadata.get(
                    "source",
                    LAW_FILE_PATH,
                ),
                "article_number": article_number,
                "article_title": article_title,
                "future_effective_note": "[시행일:" in article_content,
            },
        )

        article_documents.append(article_document)

    return article_documents
```

```python
article_documents = []

for document in documents:
    article_documents.extend(
        split_law_articles(document)
    )

print(f"조문 단위 문서 수: {len(article_documents)}")

for document in article_documents[:3]:
    print("-" * 80)
    print("조문:", document.metadata["article_number"])
    print("제목:", document.metadata["article_title"])
    print(document.page_content[:300])
```

```python
text_splitter = RecursiveCharacterTextSplitter(
    chunk_size=1500,
    chunk_overlap=200,
    separators=[
        "\n\n",
        "\n",
        "①",
        "②",
        "③",
        "④",
        "⑤",
        ". ",
        " ",
    ],
)

law_documents = text_splitter.split_documents(
    article_documents
)

print(f"최종 분할 문서 수: {len(law_documents)}")

for document in law_documents[:3]:
    print("-" * 80)
    print(document.metadata)
    print(document.page_content[:300])
```

```python
import shutil

from langchain_chroma import Chroma
from langchain_ollama import OllamaEmbeddings


LAW_DB_DIRECTORY = "./chroma_income_tax_db"

# 문서 또는 분할 기준을 변경했을 때만 True
REBUILD_DB = True

if REBUILD_DB:
    shutil.rmtree(
        LAW_DB_DIRECTORY,
        ignore_errors=True,
    )

# 기존 노트북에서 사용한 임베딩 모델
embeddings_model = OllamaEmbeddings(
    model="bge-m3"
)

# Chroma 벡터 저장소 생성
law_db = Chroma.from_documents(
    documents=law_documents,
    embedding=embeddings_model,
    collection_name="income_tax_law_20260701",
    persist_directory=LAW_DB_DIRECTORY,
)

print("소득세법 벡터 저장소 생성 완료")
```

```python
law_retriever = law_db.as_retriever(
    search_kwargs={
        "k": 4,
    }
)
```

```python
from langchain_core.tools import tool


@tool
def search_income_tax_law(query: str) -> str:
    """
    2026년 7월 1일 시행 소득세법 문서에서 관련 조문을 검색합니다.

    소득 종류, 비과세소득, 소득공제, 세액공제, 세율,
    납세의무, 원천징수 및 소득세법 조문에 관한 질문에 사용합니다.
    """

    docs = law_db.similarity_search(
        query,
        k=4,
    )

    if not docs:
        return (
            "첨부된 소득세법에서 "
            "관련 조문을 찾지 못했습니다."
        )

    formatted_results = []

    for index, doc in enumerate(docs, start=1):
        article_number = doc.metadata.get(
            "article_number",
            "조문 번호 확인 불가",
        )

        article_title = doc.metadata.get(
            "article_title",
            "",
        )

        result = f"""
[검색 결과 {index}]
조문: {article_number}
제목: {article_title}
미래 시행 관련 문구 포함: {doc.metadata.get("future_effective_note", False)}

{doc.page_content}
""".strip()

        formatted_results.append(result)

    return "\n\n" + "\n\n---\n\n".join(
        formatted_results
    )
```
